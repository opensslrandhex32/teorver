## Задача

Случайно выбирается точка внутри квадрата со стороной bigSquare=10.

В большом квадрате расположен меньший квадрат со стороной smallSquare=4.

Найти вероятность того, что случайно выбранная точка попадёт во внутренний квадрат, а затем проверить результат с помощью компьютерного моделирования.

```python
import numpy as np
import matplotlib.pyplot as plt

N = 10000
bigSquare = 10
smallSquare = 4
# Случайные точки внутри большого квадрата
x = np.random.uniform(0, bigSquare, N)
y = np.random.uniform(0, bigSquare, N)

# Маленький квадрат находится от smallSquare - 1 до smallSquare * 2 - 1 по обеим координатам
inside = (x >= smallSquare - 1) & (x <= smallSquare * 2 - 1) & (y >= smallSquare - 1) & (y <= smallSquare * 2 - 1)

frequency = np.sum(inside) / N

print(f"Теоретическая вероятность: 0.16")
print(f"Экспериментальная частота: {frequency:.4f}")
```

```python
plt.figure(figsize=(6, 6))

plt.scatter(x[~inside], y[~inside], s=bigSquare // 2, label="Промах")
plt.scatter(x[inside], y[inside], s=bigSquare // 2, label="Попадание")

plt.xlim(0, bigSquare)
plt.ylim(0, bigSquare)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Моделирование геометрической вероятности")
plt.legend()
plt.grid(True)

plt.show()
```

### Вывод

Теоретическая вероятность попадания точки во внутренний квадрат равна **0.16**.
В результате моделирования экспериментальная частота получается близкой к этому значению. При увеличении числа испытаний она будет стремиться к теоретической вероятности, т.е. моделирование подтверждает закон больших чисел.
