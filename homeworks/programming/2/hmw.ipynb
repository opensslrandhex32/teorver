### Задача 1.38.
> Пятитомное собрание сочинений расположено на полке в случайном порядке. Какова вероятность того, что книги стоят слева направо в порядке нумерации томов (от 1 до 5)?

$P(A) = \frac{m}{n} = \frac{1}{P_5} = \frac{1}{5!} = \frac{1}{120}$

```python
display(1/120, 1/sc.factorial(5))
```

### Задача 1.43.
> Наудачу взятый телефонный номер состоит из 5 цифр. Известно, что номер телефона не начинается с цифры ноль. Какова вероятность того, что в нем все цифры:

> а) различные;

$P(A) = \frac{m}{n} = \frac{1}{A_{10}^5\cdot0,9} = \frac{1}{27216}$

```python
display(1/27216, 1/(sc.perm(10, 5)*0.9))
```

> б) одинаковые;

$P(B) = \frac{m}{n} = \frac{9}{9\cdot10^4} = 0,0001$

```python
display(9/(9*10**4))
```

> в) нечетные?

$P(C) = \frac{m}{n} = \frac{5^5}{9\cdot10^4} = \frac{5}{144}$

```python
display(5**5/(9*10**4))
```

### Задача 1.48.
> В старинной игре в кости необходимо было для выигрыша получить при бросании трех игральных костей сумму очков, превосходящую 10. Найти вероятности:

```python
from itertools import  product

dice = [1, 2, 3, 4, 5, 6]

all = list(product(dice, repeat=3))

points_11 = [1 for points in all if sum(points) == 11]
points_win = [1 for points in all if sum(points) > 10]

display(len(all), len(points_11), len(points_win))
```

> а) выпадения 11 очков;

$P(A) = \frac{27}{216} = \frac{1}{8} = 0,125$

```python
display(27/216)
```

> б) выигрыша.

$P(B) = \frac{108}{216} = \frac{1}{2} = 0,5$

```python
display(108/216)
```
