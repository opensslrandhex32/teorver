## Пример 1.8

В конкурсе по 5 номинациям участвуют 10 кинофильмов. Сколько существует вариантов распределения призов, если по каждой номинации установлены:
- а) различные призы;
- б) одинаковые призы?

### Решение

**а)** Каждый вариант распределения призов — это выбор 5 фильмов из 10 с учетом порядка по номинациям. Один и тот же фильм может получить несколько призов.

Это размещения с повторениями:

$$\overline{A}_{10}^{5}=10^5.$$

```python
# а)
A = 10**5
A
```

Получаем $100000$ вариантов.

**б)** Если призы одинаковые, порядок следования фильмов по номинациям значения не имеет. Получаем сочетания с повторениями из 10 элементов по 5:

$$\widetilde{C}_{10}^{5}=C_{10+5-1}^{5}=C_{14}^{5}.$$

```python
# б)
B = sc.comb(14, 5)
B
```

Получаем $2002$ варианта.



## Пример 1.9

Сколько существует семизначных чисел, состоящих из цифр 4, 5 и 6, в которых цифра 4 повторяется 3 раза, а цифры 5 и 6 — по 2 раза?

### Решение

Каждое такое число отличается от другого порядком следования цифр и является перестановкой с повторениями из 7 элементов.

Здесь $n_1=3$, $n_2=2$, $n_3=2$, поэтому

$$P_7(3;2;2)=\frac{7!}{3!\,2!\,2!}.$$

```python
N = sc.factorial(7) // (sc.factorial(3) * sc.factorial(2) * sc.factorial(2))
N
```

## Пример 1.10

Буквы Т, Е, И, Я, Р, О написаны на отдельных карточках. Ребенок берет карточки в случайном порядке и прикладывает одну к другой: а) 3 карточки; б) все 6 карточек. Какова вероятность того, что получится слово: а) «ТОР»; б) «ТЕОРИЯ»?

### Решение

**а)** Пусть событие $A$ — получение слова «ТОР».

Различные комбинации трех букв представляют размещения: они отличаются и составом букв, и порядком их следования. Поэтому общее число случаев

$$n=\overline{A}_6^3=6\cdot5\cdot4.$$ 

Благоприятствует событию $A$ только один случай. Тогда

$$P(A)=\frac{1}{\overline{A}_6^3}=\frac{1}{6\cdot5\cdot4}=\frac{1}{120}.$$

```python
P_A = 1 / (6 * 5 * 4)
P_A
```

**б)** Пусть событие $B$ — получение слова «ТЕОРИЯ».

Здесь используются все 6 букв, поэтому общее число случаев — перестановки из 6 элементов:

$$n=P_6=6!.$$ 

Благоприятствует событию $B$ только один случай. Поэтому

$$P(B)=\frac{1}{P_6}=\frac{1}{6!}=\frac{1}{720}.$$

```python
P_B = 1 / sc.factorial(6)
P_B
```

## Пример 1.11

Используя условие примера 1.10, найти вероятность того, что получится слово «АНАНАС», если на отдельных карточках написаны три буквы А, две буквы Н и одна буква С.

### Решение

Пусть событие $B$ — получение слова «АНАНАС».

Как и в примере 1.10, общее число случаев

$$n=P_6=6!.$$ 

Число благоприятствующих случаев больше одного: перестановка трех букв А может осуществляться $P_3=3!$ способами, а перестановка двух букв Н — $P_2=2!$ способами. По правилу произведения

$$m=P_3\cdot P_2.$$ 

Поэтому

$$P(B)=\frac{m}{n}=\frac{P_3P_2}{P_6}=\frac{3!\,2!}{6!}=\frac{1}{60}.$$

```python
P = sc.factorial(3) * sc.factorial(2) / sc.factorial(6)
P
```

## Пример 1.16

За круглым столом рассаживаются 5 мужчин и 5 женщин. Найти вероятность того, что:
- а) никакие два лица одного пола не сидят рядом;
- б) супруги сидят рядом, если эти мужчины и женщины образуют 5 супружеских пар.

### Решение

**а)** Пусть событие $A$ — никакие два лица одного пола не сидят рядом.

Общее число способов рассадки 10 лиц на 10 мест:

$$n=P_{10}=10!.$$ 

Если женщины займут четные места, это можно сделать $5!$ способами, а мужчины займут нечетные места также $5!$ способами. Можно рассадить и наоборот.

Поэтому число благоприятствующих событию $A$ случаев:

$$m_1=2(5!)^2.$$ 

Тогда

$$P(A)=\frac{m_1}{n}=\frac{2(5!)^2}{10!}=0{,}00794.$$

```python
P = 2 * sc.factorial(5)**2 / sc.factorial(10)
P
```

**б)** Пусть событие $B$ — супруги, образующие пять супружеских пар, сидят рядом.

Сначала рассадим 5 супружеских пар как 5 объектов. Это можно сделать $5!$ способами. В каждой паре возможна перестановка мужа и жены, поэтому ещё $2^5$ способов.

Получаем

$$m_2=5!\,2^5.$$ 


$$P(B)=\frac{m_2}{n}=\frac{5!\,2^5}{10!}=0{,}00106.$$

```python
P = sc.factorial(5) * 2**5 / sc.factorial(10)
P
```

## Пример 1.17

В купейный вагон (9 купе по 4 места) семи пассажирам продано 7 билетов. Найти вероятности того, что пассажиры попали: а) в два купе; б) в семь купе; в) в три купе.

### Решение

Всего в вагоне $9\cdot4=36$ мест. Общее число способов выбора 7 мест из 36:

$$n=C_{36}^{7}.$$

```python
n = sc.comb(36, 7)
n
```

**а)** Пусть событие $A$ — пассажиры попали в два купе.

Два купе из 9 можно выбрать $C_9^2$ способами. Затем 7 мест из имеющихся в этих двух купе (всего 8 мест) можно выбрать $C_8^7$ способами.

По правилу произведения

$$m_1=C_9^2\cdot C_8^7.$$ 

Поэтому

$$P(A)=\frac{C_9^2C_8^7}{C_{36}^7}.$$


```python
P_a = sc.comb(9, 2) * sc.comb(8, 7) / sc.comb(36, 7)
P_a
```

**б)** Пусть событие $B$ — пассажиры попали в семь купе.

7 купе из 9 можно выбрать $C_9^7$ способами. Семь мест в семи купе можно получить, если в каждом купе выбрать по одному месту. Это возможно $4^7$ способами.

По правилу произведения

$$m_2=C_9^7\cdot4^7.$$ 

Итак,

$$P(B)=\frac{C_9^7\,4^7}{C_{36}^7}.$$

```python
P_b = sc.comb(9, 7) * 4**7 / sc.comb(36, 7)
P_b
```

**в)** Пусть событие $D$ — пассажиры попали в три купе.

3 купе из 9 можно выбрать $C_9^3$ способами.

Нужно распределить 7 мест между тремя выбранными купе. Возможны варианты по количеству мест в купе:

$$4+2+1,\quad 3+3+1,\quad 3+2+2.$$ 

С учетом перестановок купе таких вариантов будет соответственно 6, 3 и 3.

Для этих трех случаев число способов выбора мест:

$$C_4^4C_4^2C_4^1,\quad C_4^3C_4^3C_4^1,\quad C_4^3C_4^2C_4^2.$$ 

Поэтому

$$m_3=C_9^3\left(6C_4^4C_4^2C_4^1+3C_4^3C_4^3C_4^1+3C_4^3C_4^2C_4^2\right).$$

```python
m3 = sc.comb(9, 3) * (
    6 * sc.comb(4, 4) * sc.comb(4, 2) * sc.comb(4, 1)
    + 3 * sc.comb(4, 3) * sc.comb(4, 3) * sc.comb(4, 1)
    + 3 * sc.comb(4, 3) * sc.comb(4, 2) * sc.comb(4, 2)
)
m3
```

Получаем

$$m_3=84(6\cdot24+3\cdot64+3\cdot144)=84\cdot768=64512.$$ 

Следовательно,

$$P(D)=\frac{m_3}{C_{36}^7}=\frac{64512}{8347680}\approx0{,}00773.$$

```python
P_c = m3 / sc.comb(36, 7)
P_c
```
